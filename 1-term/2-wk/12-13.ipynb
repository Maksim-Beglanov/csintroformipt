```python
import numpy as np
from numpy import linalg as LA
import matplotlib

matplotlib.use('TkAgg')
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(nrows=3, ncols=3, figsize=(8, 6))

def lissajous(a, b, delta = 0, num=2000, t_max=2 * np.pi):
    t = np.linspace(0, t_max, num)
    x = np.sin(a * t + delta)
    y = np.sin(b * t)
    return x, y

def seg_intersect(p1, p2, p3, p4):
    x1, y1 = p1
    x2, y2 = p2
    x3, y3 = p3
    x4, y4 = p4
    # Get equation for the first line:
    # p1 + t(p2 - p1) = P(t), 0 <= t <= 1
    # p3 + u(p4 - p3) = Q(u), 0 <= u <= 1
    # Get system:
    # x1 + t(x2 - x1) = x3 + u(x4 - x3)
    # y1 + t(y2 - y1) = y3 + u(y4 - y3)
    # Finally matrix:
    # t(x2 - x1) -u(x4 - x3) = x3 - x1
    # t(y2 - y1) -u(y4 - y3) = y3 - y1
    d = (x4 - x3) * (y2 - y1) - (x2 - x1) * (y4 - y3)
    if abs(d) < 1e-12:
        return None
    t = ((x4 - x3) * (y3 - y1) - (x3 - x1) * (y4 - y3))/d
    u = ((x2 - x1) * (y3 - y1) - (x3 - x1) * (y2 - y1))/d
    if 0 <= t <= 1 and 0 <= u <= 1:
        x = x1 + t * (x2 - x1)
        y = y1 + u * (y2 - y1)
        return (x, y)
    return None
 
def find_intersec(x, y, min_gap=3):
    n = len(x)
    pts = []
    for i in range(n):
        p1 = (x[i], y[i])
        p2 = (x[(i + 1) % n], y[(i + 1) % n])
        for j in range(i + min_gap, n):
            if (j + 1) % n == i:  # соседний отрезок "с другой стороны"
                continue
            p3 = (x[j], y[j])
            p4 = (x[(j + 1) % n], y[(j + 1) % n])
            pt = seg_intersect(p1, p2, p3, p4)
            if pt is not None:
                pts.append(pt)
    return np.array(pts)

for i in range(3):
    for j in range(3):
        x, y = lissajous((i+1), (j+1))
        ax[i, j].plot(x, y)
        ax[i, j].grid()

plt.show()
```

```python
a, b, delta = 5, 4, np.pi / 2
x, y = lissajous(a, b, delta, num=1200)
 
inter = find_intersec(x, y)
print(f"Найдено точек самопересечения: {len(inter)}")
 
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
 
axes[0].plot(x, y, linewidth=1, color="tab:blue")
if len(inter):
    axes[0].scatter(inter[:, 0], inter[:, 1], color="red", s=15, zorder=5)
 
if len(inter):
    axes[1].hist(inter[:, 0], bins=20, color="green", edgecolor="black")
    axes[1].set_title("Распределение X координат\nточек самопересечения")
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("количество")
 
    axes[2].hist(inter[:, 1], bins=20, color="orange", edgecolor="black")
    axes[2].set_title("Распределение Y координат\nточек самопересечения")
    axes[2].set_xlabel("y")
    axes[2].set_ylabel("количество")

plt.show()
```
